```python
a=7*10**(-4)
b=3
c=1
d=3*10**(-4)
```

```python
X, Y = np.meshgrid(np.linspace(0, 10**4, 50), np.linspace(0,10**4,50))
plt.streamplot(X, Y, (c*X - d*X*Y), (a*X*Y-b*Y) )
plt.grid()
```
