```python
Tatm=210
t0=0
T0=326
t1=60
T1=314
T2=366
```

```python
t=sp.symbols('t')
k=sp.symbols('k')
T=sp.Function('T')
```

```python
eq1= sp.Eq(
    sp.Derivative(T(t),t),
    -k*(T(t)-Tatm)
);
eq1
```

```python
eq2= sp.dsolve(eq1, ics={T(t0):T0})
eq2
```

```python
eq3=sp.Eq(
    T1,
    eq2.rhs.subs(t, t1)
)
eq3
```

```python
k_0 = sp.solveset(eq3, k, domain = sp.Reals).inf
k_0
```

```python
eq4=sp.Eq(
    T2,
    eq2.rhs.subs(k, k_0)
)
eq4
```

```python
sp.solve(eq4, t)
```
